```python
import pandas as pd

dataset_token = pd.read_csv("C:/Users/85290/OneDrive - The Chinese University of Hong Kong/stat4012/hotel_reviews_with_topics.csv")
dataset_token.head()
```

```python
import gensim.downloader as api

# Download the model and get its path
path = api.load("word2vec-google-news-300", return_path=True)
print(f"Model downloaded to: {path}")
```

```python
from gensim.models import KeyedVectors

# Load the model from the path
model_path = "C:/Users/85290/gensim-data/word2vec-google-news-300/word2vec-google-news-300.gz"
word_vectors = KeyedVectors.load_word2vec_format(model_path, binary=True)
```

```python
import ast
dataset_token['Review'] = dataset_token['Review'].apply(ast.literal_eval)
```

```python
dataset_token.head()
```

```python
review = dataset_token['Review'].tolist()
review_flattened = [element for sublist in review for element in sublist]
len(review_flattened)
```

```python
vocab = list(set(review_flattened))
vocab = [token for token in vocab if word_vectors.has_index_for(token)]
len(vocab)
```

```python
vocab = ['PAD'] + vocab
vocab = dict(zip(vocab, [i for i in range(len(vocab))]))
vocab
```

```python
# Get word vector matrix from pre-trained model
import numpy as np

vocab_keys = list(vocab.keys())

wordvec_matrix = []
for i in vocab_keys:
    wordvec_matrix.append(word_vectors.get_vector(i))

wordvec_matrix = np.array(wordvec_matrix)
wordvec_matrix
```

```python
# Convert word to index
review_indexed = [[vocab.get(element, 0) for element in row if element in vocab] for row in review]
review_indexed[:3]
```

```python
# Pads input sequences to the same length
from keras.utils import pad_sequences

review_indexed_padded = pad_sequences(review_indexed, maxlen=512, padding='post')
review_indexed_padded[:5]
```

```python
dataset_token['review_indexed_padded']=review_indexed
dataset_token.head()
```

```python
dataset_token.to_csv("hotel_reviews_with_topics_indexed.csv", index=False)
```

```python
import ast
import pandas as pd

# Assuming you've loaded the data into a DataFrame (df)

# Check if the 'review_indexed_padded' column is a string or list
def safe_convert(value):
    try:
        # If the value is a string representation of a list, convert it to a list
        if isinstance(value, str):
            return ast.literal_eval(value)
        elif isinstance(value, list):
            return value
        else:
            return None  # Return None if it's neither a string nor a list
    except Exception as e:
        print(f"Error converting value: {value}, Error: {e}")
        return None

# Apply the conversion function
df['review_indexed_padded'] = df['review_indexed_padded'].apply(safe_convert)

# Check if the column has been correctly converted
print(df['review_indexed_padded'].head())
```
